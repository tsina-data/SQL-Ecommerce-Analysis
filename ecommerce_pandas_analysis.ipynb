```python
import pandas as pd

data = {
    'customer_name' : ['Abebe', 'Aster', 'Kebede', 'Chaltu'],
    'total_spent' :[200, 300, 150, 200]
}
df = pd.DataFrame(data)
print(df)
vip_customers = df[df['total_spent'] > 200]
print(vip_customers)
```

```python
vip_customers = df[df['total_spent'] > 200]
print(vip_customers)
```

```python
total_revenue = df['total_spent'].sum()
average_spent = df['total_spent'].mean()

print("Total_Revenue: ",total_revenue)
print("Average_Spent: ",average_spent)
```
